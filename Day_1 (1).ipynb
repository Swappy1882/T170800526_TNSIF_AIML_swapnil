```python
class Dog:
    species = "Canine"   # Class Attribute

    def __init__(self, name, age):
        self.name = name      # Instance Attribute
        self.age = age        # Instance Attribute


# Creating objects of Dog class
dog1 = Dog("German", 3)
dog2 = Dog("Golden", 2)

print(dog1.name, dog1.age, dog1.species)
print(dog2.name, dog2.age, dog2.species)
```

```python
class Dog:
    species = "Canine"  # Class attribute

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def animal_name(self, dog, cat):
        self.dog = dog
        self.cat = cat
        print("Dog name:", self.dog)
        print("Cat name:", self.cat)
        print("Dog name of init fun ", self.name)

    def location(self, state, country):
        self.state = state
        self.country = country
        print("State:", self.state)
        print("Country:", self.country)



# Taking input from user
name = input("Enter dog object name: ")
age = int(input("Enter age: "))

dog_value = input("Enter dog name: ")
cat_value = input("Enter cat name: ")

# Creating object
d = Dog(name, age)
print (d.name)
print (d.age)
print (d.species)
# Calling function with user input
d.animal_name(dog_value, cat_value)

d.location("Maharashtra", "India")

```

```python
class car:
  def __init__(self, brand):
    self.brand = brand

  def show(self, colour):
    self.colour = colour
    print("Car colour : ", self.colour)
    print("Car Brand : ", self.brand)

  def display(self):
     print("Car colour : ", self.colour)
     print("Car Brand : ", self.brand )

c= car("Toyota")
c.show("Red")
c.display()
```
